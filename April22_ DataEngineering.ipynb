# Count difference
# explain difference between count(*), count(1) and COUNT(column)

count(*) - count all rows, include Null
count(1) - count rows, same as count(* ) 
count(column) - count non- null, ignore NULL


# Find the total salary per department.

```python
%sql
select department, sum(salary) AS total_salary
from emp
group by department;
```

# Count employees in each department

```python
%sql
select department, count(*) AS emp_count
from emp
GROUP BY department;
```

# find highest salary per department and sorted.

```python
%sql
select department, MAX(salary) AS max_salary
from emp
GROUP BY department
ORDER BY max_salary DESC;
```

# average salary per department

```python
%sql
select department, AVG(salary) AS avg_salary
from emp
GROUP BY department;
```

## Question: query to find the top 3 highest employees in each department

select *
from employees
order by salary desc
limit 3;

- It will give top 3 overall but not per dept,
we need:
rankinf with each dept -> then filter top 3


Window function() - over(partition by order by column name)
Lead and Lag - time - series window



```python
%sql
select *,
row_number() over(partition by department order by salary desc) as rank
from employees;
```

```python
%sql
--filter top 3 -- 
select *
from (
  select *,
  row_number() over(partition by department order by salary desc) as rank
  from employees
) temp
Where rank <= 3;

-- inner query - assign rank per department
-- outer query - filter top 3

```
